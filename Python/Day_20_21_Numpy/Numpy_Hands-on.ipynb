```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])
 
print(a + b)   # Addition       -> [11 22 33 44]
print(a - b)   # Subtraction    -> [ 9 18 27 36]
print(a * b)   # Multiplication -> [10 40 90 160]
print(a / b)   # Division       -> [10. 10. 10. 10.]
print(a % b)   # Modulus        -> [0 0 0 0]
print(a ** 2)  # Exponent       -> [100 400 900 1600]
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([4, 3, 2, 1])
print(a > b)     # [False False  True  True]
print(a == b)    # [False False False False]
print(a <= 2)    # [ True  True False False]
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([4, 3, 2, 1])
print(np.add(a, b))
print(np.subtract(a, b))
print(np.multiply(a, b))
print(np.divide(a, b))
print(np.power(a, 2))
print(np.mod(a, b))
print(np.abs(np.array([-1, -2, 3])))   # [1 2 3]
```

```python
a = np.array([10, 20, 30, 40, 50])
print(a[0])     # 10 (first element)
print(a[3])     # 40 (fourth element)
 
b = np.array([[1, 2, 3], [4, 5, 6]])
print(b[1, 2])  # 6  (row 1, column 2)
```

```python
a = np.array([10, 20, 30, 40, 50])
print(a[1:4])    # [20 30 40]
print(a[:3])     # [10 20 30]
print(a[2:])     # [30 40 50]
```

```python
a = np.arange(10)
print(a[::2])    # [0 2 4 6 8]   every 2nd element
print(a[1:8:2])  # [1 3 5 7]
print(a[::-1])   # reverse the array
```

```python
a = np.array([10, 15, 20, 25, 30])
print(a)
mask = a > 18
print(mask)        # [False False  True  True  True]
print(a[mask])     # [20 25 30]
print(a[a % 10 == 0])   # [10 20 30]
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a)
print(b)
print(np.concatenate([a, b]))    # [1 2 3 4 5 6]
print(np.vstack([a, b]))         # stack as rows -> shape (2,3)
print(np.hstack([a, b]))          # stack horizontally -> [1 2 3 4 5 6]
print(np.stack([a, b], axis=0))   # new axis -> shape (2,3)
```

```python
import time
import numpy as np
 
size = 1_000_000
py_list = list(range(size))
np_array = np.arange(size)
 
start = time.time()
sum([x * 2 for x in py_list])
print('Python list:', time.time() - start, 'seconds')
 
start = time.time()
np_array * 2
print('NumPy array:', time.time() - start, 'seconds')
```
